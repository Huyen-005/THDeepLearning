```python
import tensorflow as tf
import numpy as np

print("===== TENSORFLOW BASIC =====")

#  BIỂU DIỄN TENSOR (RANK)
print("\n RANK ")

# Scalar (Rank 0)
r0 = tf.constant(4)
print("Scalar:", r0)

# Vector (Rank 1)
r1 = tf.constant([2.0, 3.0, 4.0])
print("Vector:", r1)

# Matrix (Rank 2)
r2 = tf.constant([[1, 2],
                  [3, 4],
                  [5, 6]])
print("Matrix:\n", r2)

# Tensor 3D (Rank 3)
r3 = tf.constant(np.arange(30).reshape(3, 2, 5))
print("Tensor 3D shape:", r3.shape)


#  THUỘC TÍNH
print("\n===== ATTRIBUTES =====")
print("Shape:", r2.shape)
print("Dtype:", r2.dtype)


#  HÀM TẠO TENSOR
print("\n===== CREATE TENSOR =====")

print("Zeros:\n", tf.zeros([2,3]))
print("Ones:\n", tf.ones([2,3]))
print("Fill:\n", tf.fill([2,2], 7))

print("Linspace:\n", tf.linspace(5.0, 20.0, 5))
print("Range:\n", tf.range(0, 10, 2))


#  RANDOM TENSOR
print("\n===== RANDOM =====")

print("Normal:\n", tf.random.normal([2,3]))
print("Truncated Normal:\n", tf.random.truncated_normal([3,2,2]))
print("Uniform:\n", tf.random.uniform([2,3]))

#  PHÉP TOÁN
print("\n===== OPERATIONS =====")

a = tf.constant([1.0, 4.0])
b = tf.constant([2.0, 3.0])

print("Add:", tf.add(a,b))
print("Subtract:", tf.subtract(a,b))
print("Multiply:", tf.multiply(a,b))
print("Divide:", tf.divide(a,b))

#  HÀM TOÁN HỌC

print("\n===== MATH =====")

x = tf.constant([-4.0, 9.0])

print("Abs:", tf.abs(x))
print("Sqrt:", tf.sqrt(tf.abs(x)))
print("Round:", tf.round(tf.constant([1.2, 2.7])))
print("Ceil:", tf.math.ceil(tf.constant([1.2, 2.7])))
print("Floor:", tf.math.floor(tf.constant([1.2, 2.7])))

#  MA TRẬN

print("\n===== MATRIX =====")

A = tf.constant([[1,2],[3,4]])
B = tf.constant([[5,6],[7,8]])

print("Matmul:\n", tf.matmul(A,B))
print("Transpose:\n", tf.transpose(A))
print("Eye:\n", tf.eye(3))
print("Diag:\n", tf.linalg.diag([1,2,3]))


#  VARIABLE

print("\n===== VARIABLE =====")

v = tf.Variable(10)
print("Initial:", v)

v.assign(20)
print("Assign:", v)

v.assign_add(5)
print("Add:", v)

v.assign_sub(3)
print("Sub:", v)

#  GRAPH
print("\n===== GRAPH =====")

@tf.function
def compute(a, b):
    return a + b * b

print("Graph result:", compute(3, 5))
```

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
# 1. Tensor [2, 3] gán 0
print("Bài 1: Tensor zeros [2, 3]")
print(tf.zeros([2, 3]))

# 2. Tensor có shape/dtype như X nhưng gán 0
print("\nBài 2: zeros_like X")
X = tf.constant([[1, 2, 3], [4, 5, 6]], dtype=tf.float32)
print(tf.zeros_like(X))

# 3. Tensor [3, 2] gán 5
print("\nBài 3: Tensor fill 5 kích thước [3, 2]")
print(tf.fill([3, 2], 5))

# 4. Tensor 1-chiều 50 phần tử chẵn trong khoảng [5, 10]
print("\nBài 4: Linspace 50 phần tử từ 6.0 đến 10.0")
print(tf.linspace(6.0, 10.0, 50))

# 5. Tensor [10, 12, ..., 100]
print("\nBài 5: Range từ 10 đến 100 bước nhảy 2")
print(tf.range(10, 102, 2))

# 6. Normal distribution (mean=0, std=2)
print("\nBài 6: Random Normal (mean=0, std=2)")
print(tf.random.normal([3, 2], mean=0, stddev=2))

# 7. Uniform distribution [0, 2]
print("\nBài 7: Random Uniform trong khoảng [0, 2]")
print(tf.random.uniform([3, 2], minval=0, maxval=2))

# 8. Shuffle tensor theo hàng
print("\nBài 8: Shuffle tensor X8")
X8 = tf.constant([[1, 2], [3, 4], [5, 6]])
print(tf.random.shuffle(X8))

# 9. Random crop
print("\nBài 9: Random crop từ (10,10,3) xuống (5,5,3)")
t9_big = tf.random.uniform([10, 10, 3])
t9_crop = tf.image.random_crop(t9_big, size=[5, 5, 3])
print("Shape sau khi crop:", t9_crop.shape)
# 10. Logic: x + y nếu x < y, ngược lại x - y
print("\nBài 10: Phép toán điều kiện tf.where")
x10 = tf.random.uniform([5])
y10 = tf.random.uniform([5])
res10 = tf.where(x10 < y10, x10 + y10, x10 - y10)
print("Kết quả logic x, y:", res10)

# 11. Đường chéo và phép cộng
print("\nBài 11: Lấy đường chéo và cộng ma trận")
x11 = tf.random.uniform([3, 2], 5, 10)
y11 = tf.random.uniform([2, 3], 15, 20)
# Lưu ý: x + y chỉ thực hiện được khi khớp shape, ở đây ta transpose y
print("x + y.T:\n", x11 + tf.transpose(y11))

# 12. Hàm tính toán trong Graph
print("\nBài 12: Cộng, trừ, nhân, chia trong @tf.function")
@tf.function
def graph_math(x, y):
    return x + y, x - y, x * y, x / y

x12 = tf.constant([10.0, 20.0])
y12 = tf.constant([2.0, 5.0])
print("Kết quả (Add, Sub, Mul, Div):", graph_math(x12, y12))
# 13, 14. Model tính toán cơ bản
@tf.function
def model_basic(x, y):
    return x + y, x - y, x * y

# 15. Chu vi hình chữ nhật P = (a + b) * 2
@tf.function
def model_chu_vi(a, b):
    return tf.multiply(tf.add(a, b), 2.0)

# 16. Diện tích toàn phần hình hộp chữ nhật
@tf.function
def model_sttp(a, b, c):
    s_xq = (a + b) * c * 2
    s_tp = (a * b) * 2 + s_xq
    return s_tp

print("\nBài 15: Chu vi (a=5, b=3) =", model_chu_vi(5.0, 3.0).numpy())
print("Bài 16: Diện tích TP (a=5, b=3, c=4) =", model_sttp(5.0, 3.0, 4.0).numpy())
# 17, 18. Logic chứng minh hình thang
@tf.function
def model_hinh_thang(AD, BC, AB, CD, is_parallel):
    # Chứng minh song song hoặc bằng nhau dựa trên input
    check_parallel = tf.logical_and(is_parallel, tf.equal(AD, BC))
    return check_parallel

# 19. Chứng minh đường trung bình (NA = NC)
@tf.function
def model_midpoint_proof(AM, MB, is_MN_parallel_BC):
    # Nếu M là trung điểm AB (AM=MB) và MN // BC thì N là trung điểm AC (NA=NC)
    m_is_midpoint = tf.equal(AM, MB)
    can_conclude_n_midpoint = tf.logical_and(m_is_midpoint, is_MN_parallel_BC)
    return can_conclude_n_midpoint

print("\nBài 19: Chứng minh NA = NC")
ans19 = model_midpoint_proof(tf.constant(5.0), tf.constant(5.0), tf.constant(True))
print("Kết luận N là trung điểm AC:", "Đúng" if ans19 else "Sai")
print("\n" + "="*20 + " ĐANG KHỞI TẠO BIỂU ĐỒ " + "="*20)

# Cấu hình khung hình tổng quát (2 hàng, 2 cột)
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
plt.subplots_adjust(hspace=0.4)

# --- BÀI 6 & 7: SO SÁNH PHÂN PHỐI NGẪU NHIÊN ---
# Tạo tập dữ liệu lớn hơn để biểu đồ mượt (1000 phần tử)
normal_sample = tf.random.normal([1000], mean=0, stddev=2).numpy()
uniform_sample = tf.random.uniform([1000], minval=0, maxval=2).numpy()

axes[0, 0].hist(normal_sample, bins=30, alpha=0.6, color='blue', label='Normal (B6)', edgecolor='black')
axes[0, 0].hist(uniform_sample, bins=30, alpha=0.6, color='orange', label='Uniform (B7)', edgecolor='black')
axes[0, 0].set_title("Bài 6 & 7: Phân phối Normal vs Uniform")
axes[0, 0].legend()
axes[0, 0].grid(axis='y', alpha=0.3)

# --- BÀI 9: MINH HỌA RANDOM CROP ---
# Giả lập một tấm ảnh nhiễu 10x10 và vùng cắt 5x5
big_img = tf.random.uniform([10, 10, 3]).numpy()
# Giả định lấy vùng trung tâm để minh họa trực quan
crop_img = big_img[2:7, 2:7, :]

axes[0, 1].imshow(big_img)
axes[0, 1].set_title("Bài 9: Tensor Gốc (10x10x3)")
# Vẽ một khung đỏ để chỉ vùng sẽ bị cắt
rect = plt.Rectangle((2, 2), 5, 5, linewidth=2, edgecolor='r', facecolor='none')
axes[0, 1].add_patch(rect)

#  BÀI 15 & 16: KẾT QUẢ TÍNH TOÁN HÌNH HỌC
labels = ['Chu vi (B15)', 'Diện tích TP (B16)']
values = [model_chu_vi(5.0, 3.0).numpy(), model_sttp(5.0, 3.0, 4.0).numpy()]

axes[1, 0].bar(labels, values, color=['teal', 'coral'])
axes[1, 0].set_title("Bài 15 & 16: Kết quả Model Hình học")
for i, v in enumerate(values):
    axes[1, 0].text(i, v + 2, str(v), ha='center', fontweight='bold')

# --- BÀI 19: LOGIC ĐƯỜNG TRUNG BÌNH (Sơ đồ tư duy) ---
axes[1, 1].axis('off')
axes[1, 1].set_title("Bài 19: Logic Chứng minh Hình học")
logic_text = (
    "GIẢ THUYẾT:\n"
    "1. M là trung điểm AB (AM=MB): ĐÚNG\n"
    "2. MN song song BC (MN//BC): ĐÚNG\n\n"
    "==> KẾT LUẬN:\n"
    "N là trung điểm AC (NA=NC): ĐÚNG"
)
axes[1, 1].text(0.5, 0.5, logic_text, ha='center', va='center',
                bbox=dict(boxstyle="round", facecolor='wheat', alpha=0.5), fontsize=12)

plt.suptitle("TRỰC QUAN HÓA KẾT QUẢ THỰC HÀNH TENSORFLOW", fontsize=16, fontweight='bold')
plt.show()

print("="*20 + " HOÀN TẤT VẼ HÌNH " + "="*20)
```
